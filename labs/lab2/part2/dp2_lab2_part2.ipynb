#### 3.3.2 Диаграммы размаха и поиск выбросов

Для обнаружения статистических выбросов используются диаграммы размаха. 
Точки, расположенные за пределами усов диаграммы, потенциально являются выбросами.
При этом наличие выброса не обязательно означает ошибку в данных: экстремальное 
значение может быть реальным наблюдением.

##### Датасет мобильности транспортных потоков в умном городе

```python
for feature in features1:
    pyplot.figure(figsize=(10, 4))
    
    seaborn.boxplot(
        y=data_frame1[feature]
    )
    
    pyplot.title(f'Диаграмма размаха признака {feature}')
    pyplot.ylabel(feature)
    pyplot.grid(alpha=0.3)
    pyplot.show()
```

- `vehicle_count`: медиана ~300, широкий межквартильный размах, множество выбросов выше верхнего уса 
- `average_speed`: медиана ~40 км/ч, есть выбросы как в сторону низких (пробки), так и высоких (свободный поток) значений.
- `traffic_density`: медиана ~35, длинный верхний хвост выбросов - отдельные участки с экстремальной плотностью.
- `queue_length`: ммедиана низкая, большинство очередей короткие, но есть выбросы >100 - единичные заторы.
- `average_wait_time`: медиана низкая, но хвост выбросов значительный.
- `congestion_score`: медиана ~25, но верхний ус уходит к 100 - есть выраженная группа загруженных наблюдений.
- `emission_estimate` и `fuel_waste_estimate`: ярко выраженные экстремальные выбросы

##### Датасет стоимости такси

```python
for feature in features2:
    pyplot.figure(figsize=(10, 4))
    
    seaborn.boxplot(
        y=data_frame2[feature],
        color="red"
    )
    
    pyplot.title(f'Диаграмма размаха признака {feature}')
    pyplot.ylabel(feature)
    pyplot.grid(alpha=0.3)
    pyplot.show()
```

- `Hour_of_Day`: медиана ~12 - поездки равномерны по суткам.

## 4. Корреляционный анализ и зависимости

    - Построить матрицу корреляции для численных признаков датасета.
    - Визуализировать матрицу корреляции в виде тепловой карты (heatmap) с отображением коэффициентов.
    - Построить диаграммы рассеяния (scatterplot) или парные графики (pairplot) для пар признаков с наиболее сильной линейной зависимостью.

##### Датасет мобильности транспортных потоков в умном городе

```python
numeric_data1 = data_frame1.select_dtypes(include='number')
print('Датасет мобильности транспортных потоков в умном городе, матрица корреляции:')
correlation_matrix1 = numeric_data1.corr()
correlation_matrix1
```

##### Датасет стоимости такси

```python
numeric_data2 = data_frame2.select_dtypes(include='number')
print('Датасет стоимости такси, матрица корреляции:')
correlation_matrix2 = numeric_data2.corr()
correlation_matrix2
```

##### Датасет мобильности транспортных потоков в умном городе

```python
pyplot.figure(figsize=(18, 14))

seaborn.heatmap(
    correlation_matrix1,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0
)

pyplot.title('Датасет мобильности транспортных потоков в умном городе, матрица корреляции')
pyplot.show()
```

##### Датасет стоимости такси

```python
pyplot.figure(figsize=(12, 10))

seaborn.heatmap(
    correlation_matrix2,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0
)

pyplot.title('Матрица корреляции датасета стоимости такси')
pyplot.show()
```

##### Датасет мобильности транспортных потоков в умном городе

```python
correlation_pairs1 = (
    correlation_matrix1
    .where(
        numpy.triu(
            numpy.ones(correlation_matrix1.shape),
            k=1
        ).astype(bool)
    )
    .stack()
    .sort_values(key=abs, ascending=False)
)

print('Наиболее сильные корреляции в датасете мобильности:\n')
print(correlation_pairs1.head(10))
```

```python
strongest_pair1 = correlation_pairs1.index[0]

feature_x1 = strongest_pair1[0]
feature_y1 = strongest_pair1[1]

print('Самая сильная корреляционная пара:')
print(f'{feature_x1} и {feature_y1}')
print(
    f'Коэффициент корреляции: '
    f'{correlation_matrix1.loc[feature_x1, feature_y1]:.3f}'
)
```

```python
pyplot.figure(figsize=(8, 4))

seaborn.scatterplot(
    data=data_frame1,
    x=feature_x1,
    y=feature_y1,
    alpha=0.3
)

pyplot.title(
    f'Зависимость {feature_y1} от {feature_x1}'
)

pyplot.xlabel(feature_x1)
pyplot.ylabel(feature_y1)
pyplot.grid(alpha=0.3)
pyplot.show()
```

```python
pairplot_features1 = [
    'vehicle_count',
    'average_speed',
    'traffic_density',
    'queue_length',
    'congestion_score'
]

seaborn.pairplot(
    data_frame1[pairplot_features1],
    diag_kind='hist'
)

pyplot.suptitle(
    'Парные зависимости признаков датасета мобильности транспортных потоков в умном городе',
    y=1.02
)

pyplot.show()
```
