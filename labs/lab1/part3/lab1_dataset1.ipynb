```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path = "/home/cintie/Documents/projects/uniq/dp2/lab1/part3/datasets/faults.csv"
faults = pd.read_csv(path)

print("Размер датасета:", faults.shape)
faults.head()
```

### График

```python
fault_types = [
    "Pastry",
    "Z_Scratch",
    "K_Scatch",
    "Stains",
    "Dirtiness",
    "Bumps",
    "Other_Faults"
]

fault_counts = faults[fault_types].sum()

plt.figure(figsize=(8, 5))
fault_counts.plot(kind="bar")

plt.title("Количество записей по типам дефектов")
plt.xlabel("Тип дефекта")
plt.ylabel("Количество")
plt.show()
```
